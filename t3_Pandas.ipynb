### **Merging data frames**

 similar function on SQL "join"

```python
import pandas as pd

data1 = {
      "name": ["Sally", "Mary", "John"],
      "age": [50, 40, 30]
      }

data2 = {
    "name": ["Sally", "Peter", "Micky"],
    "age": [77, 44, 22]
    }

df1 = pd.DataFrame(data1)

df2 = pd.DataFrame(data2)
```

```python
newdfa = df1.merge(df2, how = "right")

newdfa
```

```python
newdfa = df1.merge(df2, how = "right", on = "name")

newdfa  # (auto renaming)
```

```python
newdfb = df1.merge(df2, how = "left")

newdfb
```

```python
newdfb = df1.merge(df2, how = "left", on = "name")

newdfb
```

```python
newdfc = df1.merge(df2, how = "inner")

newdfc
```

```python
newdfc = df1.merge(df2, how = "inner", on = "name")

newdfc
```

```python
newdfd = df1.merge(df2, how = "outer")

newdfd
```

```python
newdfd = df1.merge(df2, how = "outer", on = "name")

newdfd
```

```python
newdfe = df1.merge(df2, how = "cross")

newdfe      # combination of all difference row
```

```python
newdfe = df1.merge(df2, how = "cross", on = "name")

newdfe
```

```python
type(newdfe)
```

### **Pivot table**

it is multi dimensional version of GroupBy

```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

plt.style.use("ggplot")
```

```python
df = pd.read_csv("https://bit.ly/3kGDuKx", na_values = "?")

df.head()
```

```python
df.drop(["ticket", "name"], inplace = True, axis = 1)
```

```python
df.head(2)
```

```python
table = pd.pivot_table(data = df, index = ["sex"])    # avg values

table   # average values (mean)
```

```python
table = pd.pivot_table(df, index = ["sex", "pclass"])

table
```

```python
table = pd.pivot_table(df, index = ["sex", "pclass"], aggfunc = {"age": np.mean, "fare": np.mean})

table
```

```python
table = pd.pivot_table(data = df, index = ["sex", "pclass"], values = ["age"], aggfunc = np.mean)

table
```

```python
table = pd.pivot_table(data = df, index = ["sex", "pclass"], values = ["age", "fare"], aggfunc = np.mean)

table
```

```python
df.info()
```

```python
table = pd.pivot_table(df, index = ["sex", "survived", "pclass"], columns = ["embarked"], values = ["age"], aggfunc = np.mean)

table
```

```python
table = pd.pivot_table(df, index = ["sex", "survived", "pclass"], columns = ["embarked"], values = ["age"], aggfunc = np.mean, fill_value = np.mean(df["age"]))


table
```
